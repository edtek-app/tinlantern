# Model comparison — presentation only

**This notebook is NOT authoritative.** Every number here is computed by
the tested functions in `ml/evaluation/`, and the numbers that get quoted
live in `evals/reports/`. Nothing cites this file, so a stale copy cannot
mislead anyone — which is a stronger guarantee than a test asserting it
looks right.

It exists to let a human read the comparison alongside the data. If it
disagrees with `evals/reports/`, the report is correct and this is stale.


```python
from sqlalchemy import create_engine

from data.generator import load_config
from ml.evaluation import candidates, load_labels, run_comparison
from ml.src.features import extract_window_features

config = load_config("data/generator/cohort.example.yaml")
engine = create_engine(__import__("os").environ["DATABASE_URL"])
with engine.connect() as connection:
    absolute = extract_window_features(connection, config)

labels = load_labels()
comparison = run_comparison(
    absolute,
    labels["at_risk"].loc[absolute.index],
    labels["archetype"].loc[absolute.index],
    candidates(),
)
print(comparison.summary())
```

## What to read first

Per-archetype recall, with its `n`. Aggregate AUC is secondary and
flattering on this cohort: struggling learners are both highly engaged
and all at risk, so any model that catches them on scores alone posts a
strong figure. An archetype below the reportability floor is marked
unreportable rather than quoted.
